### Linear Regression

with **Ordinary Least Squares(OLS)**

- ɵ = (X<sup>T</sup>.X)<sup>-1</sup>.X<sup>T</sup>.y


```python
import numpy as np
```

```python
class LinearRegressionOLS():
    def __init__(self):
        self.bias = None
        self.weights = None

    def fit(self, X, y):
        n_samples, n_features = X.shape #number of samples(rows) and features(columns) in the dataset
        
        #STEP 1: Adding a column of ones to the feature matrix for bias(c) term
        X_updated = np.c_[np.ones(n_samples), X] 
        #STEP 2: Calculating the weights and bias using the closed-form solution
        
        X_transpose = X_updated.T #transpose of the feature matrix

        #closed-form solution for linear regression: w = (X^T * X)^(-1) * X^T * y
        thetas = np.linalg.inv(X_transpose.dot(X_updated)).dot(X_transpose).dot(y)
        # thetas = np.dot(np.linalg.inv(np.dot(X_transpose, X_updated)), np.dot(X_transpose, y))
        # thetas = np.linalg.pinv(X_transpose @ X_updated) @ X_transpose @ y #@ for matrix multiplication
        self.bias = thetas[0] #first element is the bias term
        self.weights = thetas[1:] #remaining elements are the weights for each feature

    def predict(self, X):
        #STEP 3: Making predictions using the learned weights and bias
        y_pred = self.bias + np.dot(X, self.weights) # c + m1*x1 + m2*x2 + ... + mn*xn
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10]) 
model = LinearRegressionOLS()
model.fit(X, y)
predicted = model.predict(X)
print(predicted) 
print("bias:", model.bias)
print("weights:", model.weights)
```
