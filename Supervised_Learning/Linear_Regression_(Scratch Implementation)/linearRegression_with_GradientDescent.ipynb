## Linear Regression

- y = mx + c
- y_pred = ɵ<sub>0</sub> + ɵ<sub>1</sub>.x + ɵ<sub>2</sub>.x + ɵ<sub>3</sub>.x ...


```python
import numpy as np
```

```python
class LinearRegression():
    def __init__(self, learning_rate=0.01, n_iters = 1000):
        self.bias = None #bias(c) for the linear regression model
        self.weights = None #weights(m) for the features
        self.learning_rate = learning_rate #learning rate for gradient descent
        self.n_iters = n_iters #number of iterations for training the model

    def fit(self, X, y):
        n_samples, n_features = X.shape #number of samples(rows) and features(columns) in the dataset
        #STEP 1: initialize weights and bias to zeros
        self.bias = 0
        self.weights = np.zeros(n_features) # [0, 0, ..., 0] for each feature

        # training the model for n_iters iterations
        for _ in range(self.n_iters):
            #STEP 2: Training the model using gradient descent
            y_pred = self.bias + np.dot(X, self.weights) # c + m1*x1 + m2*x2 + ... + mn*xn

            #STEP 3: Calculating the gradients
            db = (1/n_samples) * np.sum(y_pred - y) #derivative of the loss function with respect to bias
            dw = (1/n_samples) * np.dot(X.T, (y_pred - y)) #derivative of the loss function with respect to weights

            #STEP 4: Updating the weights and bias using the gradients
            self.bias -= self.learning_rate * db
            self.weights -= self.learning_rate * dw

    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights) # c + m1*x1 + m2*x2 + ... + mn*xn
        return y_pred
        

```

```python
X = np.array([[1, 3], [2,4], [3, 5], [4, 6], [5, 7]])
y = np.array([2, 4, 6, 8, 10])
regressor = LinearRegression(learning_rate=0.01, n_iters=1000)
regressor.fit(X, y)
predicted = regressor.predict(X)
print(predicted)
```
