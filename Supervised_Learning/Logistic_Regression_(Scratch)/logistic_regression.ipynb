## Logistic Regression


```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.01, n_iters = 1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iters = n_iters
    
    def _sigmoid(self, x):
        return 1 / (1 + np.exp(-x))
    
    def fit(self, X, y):
        m, n = X.shape #number of samples(rows) and features(columns) in the dataset
        self.bias = 0
        self.weights = np.zeros(n) # [0, 0, ..., 0] for each feature

        for _ in range(self.n_iters):
            z = self.bias + np.dot(X, self.weights) # linear model
            y_pred = self._sigmoid(z) # applying sigmoid function to get probabilities

            # calculating gradients
            db = (1/m) * (y_pred - y) #derivative of the loss function with respect to bias
            dw = (1/m) * np.dot(X.T, (y_pred - y)) #derivative of the loss function with respect to weights

            # updating parameters
            self.bias -= self.lr * db
            self.weights -= self.lr * dw
            
    def get_probabilities(self, x):
        linear_model = self.bias + np.dot(x, self.weights)
        return self._sigmoid(linear_model)
    
    def predict(self, X, threshold=0.5):
        y_pred_prob = self.get_probabilities(X)
        return (y_pred_prob >= threshold).astype(int)
```

```python
X = np.array([[1, 9], [2,4], [3, 5], [4, 6], [5, 7], [6,  8]])
y = np.array([0, 1, 0, 1, 1, 0]) # labels/classes
model = LogisticRegression(learning_rate=0.01, n_iters=1000)
model.fit(X, y)
predicted = model.predict(X)
print(predicted)
print("bias:", model.bias)
print("weights:", model.weights)
print("probabilities:", model.get_probabilities(X))
```
