```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(data, type(data))
```

```python
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    unique_users = set()
    cleaned_data = []
    for user in data:
        #clean ratings - data consistency
        if type(user["rating"]) != float:
            raw_rating = user["rating"].strip().lower()
            rating = ""
            if raw_rating in text_to_num:
                rating = text_to_num[raw_rating]
            else:
                rating = raw_rating
            user["rating"] = float(rating)
        #missing data->None
        raw_age = user.get("age")
        if raw_age == None:
            user["age"] = None
        #deduplication
        user_name = user['name'].strip()
        if user_name in unique_users:
            continue
        unique_users.add(user_name)
        cleaned_data.append(user)
    return cleaned_data
        
        
```

```python
cleaned_data = clean_data(data)
```

```python
def get_insights(data):
    total = 0
    counter = 0
    for user in data:
        total += user["rating"]
        if user['rating'] <3:
            counter += 1
    print(f"Average Rating of {total/len(data)}")
    print(f"{counter/len(data) * 100}% of user with poor rating")

    
```

```python
get_insights(cleaned_data)
```
