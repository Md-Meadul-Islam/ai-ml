```python
cdf = df.copy()
```

```python
cdf.ffill() # forward fill; it fills empty values from top to bottom
```

```python
cdf.bfill()# backward fill; it fills empty values from bottom to top
```

```python
avg_age = cdf["age"].mean()
cdf["age"] = cdf["age"].fillna(avg_age)
```
