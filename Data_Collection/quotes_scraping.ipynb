```python
from bs4 import BeautifulSoup
import requests
```

## fetch data and store it

```python
try:
    page = 9
    while True:
        url = f"https://quotes.toscrape.com/page/{page}/"
        res = requests.get(url)
        res.raise_for_status()
        res.raise_for_status()
        if res.status_code == 200:
            html_content = res.text
            soup = BeautifulSoup(html_content, "lxml")
            quotes = soup.select("div.quote")
            if not quotes:
                print("No more pages !")
                break;

            with open(f"scraped_data/quotes_{page}.html", "w", encoding="utf-8") as f:
                f.write(html_content)
            print(page)
            page += 1
        
    
except requests.exceptions.HTTPError as e:
    print("HTTP Error:", e)

except requests.exceptions.RequestException as e:
    print("Other Request Error:", e)
```

## Extract data / beautify data

```python
try:
    cleaned_data = []
    for i in range(1, 11):
        with open(f"scraped_data/quotes_{i}.html", "r", encoding = "utf-8") as f:
            html_content = f.read()
        soup = BeautifulSoup(html_content, "lxml")
        all_quotes = soup.select("div.quote")

        for q in all_quotes:
            all_tags  = []
            for tag in q.select(".tags .tag"):
                all_tags.append(tag.get_text())
            if "life" in all_tags:
                quote = q.select_one("span.text").get_text()
                author = q.select_one(".author").get_text()
                cleaned_data.append([quote, author])
                
except requests.exceptions.RequestException as e:
    print("Other Request Error:", e)
```

```python
import pandas as pd
df = pd.DataFrame(cleaned_data, columns=["Quotes", "Author"])
df

df.to_csv("cleaned_data/life_quotes.csv", index=False)
```

```python
try:
    cleaned_data = []
    for i in range(1, 11):
        with open(f"scraped_data/quotes_{i}.html", "r", encoding = "utf-8") as f:
            html_content = f.read()
        soup = BeautifulSoup(html_content, "lxml")
        all_quotes = soup.select("div.quote")

        for q in all_quotes:
            quote = q.select_one("span.text").get_text()
            author = q.select_one(".author").get_text()
            tags = []
            for tag in q.select(".tags .tag"):
                tags.append(tag.get_text())
            cleaned_data.append([quote, author, tags])
                
except requests.exceptions.RequestException as e:
    print("Other Request Error:", e)
```

```python
df = pd.DataFrame(cleaned_data, columns=["Quotes", "Author", "Tags"])

df.to_csv("cleaned_data/quotes.csv", index=False)
```
