```python
import requests
import os
```

# Response

- `res.status_code` - HTTP response code 
- `res.content` - Raw Bytes data
- `res.text` - HTML content of the page
- `res.headers` - MetaData of requests
  

```python
url = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(url)
res.status_code
```

```python
os.chdir("Data_Collection")
print(os.getcwd())
```

```python
if res.status_code == 200:
    os.makedirs("scraped_data", exist_ok=True)
    with open("scraped_data/data1.html", "w", encoding="utf-8") as f:
        f.write(res.text)

    
```
