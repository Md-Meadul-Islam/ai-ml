```python
import requests
import os
```

# Response

- `res.status_code` - HTTP response code 
- `res.content` - Raw Bytes data
- `res.text` - HTML content of the page
- `res.headers` - MetaData of requests
  

```python
url = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(url)
res.status_code
```

```python
os.chdir("Data_Collection")# change directory
print(os.getcwd())
os.makedirs("scraped_data", exist_ok=True) # make directory if not exists
```

```python
if res.status_code == 200:
    with open("scraped_data/data1.html", "w", encoding="utf-8") as f:
        f.write(res.text)

    
```

### TRY-EXCEPT blocks

```python
try:
    url = "https://www.scrapethissite.com/pages/simple/"
    res = requests.get(url, timeout=5)
    res.raise_for_status()
    
    if res.status_code == 200:
        with open("scraped_data/data1.html", "w", encoding="utf-8") as f:
            f.write(res.text)
            
except requests.exceptions.RequestException as e:
    print("Error: ", e)
    
    
```
