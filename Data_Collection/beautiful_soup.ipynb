```python
from bs4 import BeautifulSoup
```

```python
with open("scraped_data/data1.html", "r") as f:
    html_content = f.read()
soup = BeautifulSoup(html_content, "lxml") # ["html.parser", "lxml"]
```

```python
soup.h1 # first h1 tag in html doc
soup.find("h1") # first h1 tag in html doc
```

```python
all_h3 = soup.find_all("h3")

for h3 in all_h3:
    name = h3.get_text(strip=True)
    # population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    population = h3.find_next("div").select_one("span.country-population").get_text(strip=True)
    print(population)
    # print(h3.attrs)
```

```python
all_h3 = soup.find_all("h3")
all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    country_details = h3.find_next_sibling("div")
    
    capital = country_details.select_one("span.country-capital").get_text(strip=True)
    population = country_details.select_one("span.country-population").get_text(strip=True)
    area = country_details.select_one("span.country-area").get_text(strip=True)
    
    all_countries.append([name, capital, population,area])
```

```python
all_countries
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns=["Name", "Capital","Population", "Area"])
```

```python
df.to_csv("cleaned_data/country_data.csv", index=False)
```
