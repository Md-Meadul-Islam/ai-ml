```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-2*np.pi, 2* np.pi, 100)
y = np.sin(x)

#background
plt.style.use("seaborn-v0_8")
#Plot the graph
plt.plot(x, y, label="y = sin(x)", color="blue")

#heightlight x=0 and y=0 lines
plt.axvline(0, color="red", linestyle="--", label="x = 0")
plt.axhline(0, color="green", linestyle="--", label="y = 0")

plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.title("Graph of y = sin(x)")
plt.legend()
plt.tight_layout()
```

```python
x = np.linspace(-2*np.pi, 2* np.pi, 100)
y = np.sin(x)
y1 = 2 * np.sin(x)
y2 = np.sin(x) + 2
y3 = np.sin(x) - 2

#Plot the graph
plt.plot(x, y, label="y = sin(x)", color="orange")
plt.plot(x, y1, label="y = sin(x) * 2", color="blue")
plt.plot(x, y2, label="y = sin(x) + 2", color="red")
plt.plot(x, y3, label="y = sin(x) - 2", color="green")

#heightlight x=0 and y=0 lines
plt.axvline(0, color="black", linestyle="--", label="x = 0")
plt.axhline(0, color="black", linestyle="--", label="y = 0")

plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.title("Graph of y = [2]sin(x)")
plt.legend()
plt.tight_layout()
```

```python
x = np.linspace(-2, 2, 10)
y = np.pow(x, 2)
addition = np.add(x, 5)
y1 = np.pow(addition, 2)
subtraction = np.subtract(x, 5)
y2 = np.pow(subtraction, 2)
#Plot the graph
plt.plot(x, y, label="y = x^2", color="orange")
plt.plot(x, y1, label="y = (x+5) ^ 2", color="blue")
# plt.plot(x, y2, label="y = (x-5) ^ 2", color="red")

#heightlight x=0 and y=0 lines
plt.axvline(0, color="black", linestyle="--", label="x = 0")
plt.axhline(0, color="black", linestyle="--", label="y = 0")

plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.title("Graph of y = x^2")
plt.legend()
```
