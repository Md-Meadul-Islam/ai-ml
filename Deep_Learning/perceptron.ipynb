# Single Neuron, Perceptron

```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42)
```

```python
inputs = torch.tensor([1.0, 2.0, 3.0, 4.0])
inputs
```

```python
neuron = nn.Linear(in_features=4, out_features=1)
neuron
```

```python
output = neuron(inputs)
output
```

```python
# verify.... formula

neuron.weight @ inputs + neuron.bias
```
