```python
import numpy as np
import matplotlib.pyplot as plt

import matplotlib.ticker as ticker


fig, axs = plt.subplots(2, 1, figsize=(5.4, 5.4), layout='constrained')
print(fig, axs)
x = np.arange(100)
for nn, ax in enumerate(axs):
    ax.plot(x, x)
    if nn == 1:
        ax.set_title('Manual ticks')
        ax.set_yticks(np.arange(0, 100.1, 100/3))
        xticks = np.arange(0.50, 101, 20)
        xlabels = [f'\\${x:1.2f}' for x in xticks]
        ax.set_xticks(xticks, labels=xlabels)
    else:
        ax.set_title('Automatic ticks')
```

# 3 Bar

```python
# Data
years = [2020, 2021, 2022]

movie_a = [500, 700, 900]
movie_b = [400, 650, 850]
movie_c = [300, 600, 750]

# X positions
x = np.arange(len(years))

# Width of each bar
width = 0.2

# Create figure
plt.figure(figsize=(10, 5))

# Multi bars
plt.bar(x - width, movie_a, width, label='Movie A')
plt.bar(x,         movie_b, width, label='Movie B')
plt.bar(x + width, movie_c, width, label='Movie C')

# Labels
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.title("Movie Revenue Comparison")

# X ticks
plt.xticks(x, years)

# Legend
plt.legend()

# Grid
plt.grid(axis='y')

plt.show()
```

# 2 bar

```python
# Data
years = [2020, 2021, 2022]

movie_a = [500, 700, 900]
movie_b = [400, 650, 850]

# X positions
x = np.arange(len(years))

# Width of each bar
width = 0.2

# Create figure
plt.figure(figsize=(10, 4))

# Multi bars
plt.bar(x - width/2, movie_a, width, color=['red', 'green', 'blue'], edgecolor='black', linewidth=2, alpha=1, label='Movie A')
plt.bar(x + width/2, movie_c, width, hatch="|", label='Movie B')

# Labels
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.title("Movie Revenue Comparison")

# X ticks
plt.xticks(x, years)

# Legend
plt.legend()

# Grid
plt.grid(axis='y')

plt.show()
```

# Multi bar

```python
# Data
years = [2020, 2021, 2022]

movies = {
"Movie A" : [500, 700, 900],
"Movie B" : [400, 650, 850],
"Movie C" : [300, 600, 750],
"movie D" : [320, 900, 710],
"movie E" : [1320, 90, 1000]
    
}

# X positions
x = np.arange(len(years))

# Width of each bar
n= len(movies)
width = 0.8/n

plt.figure(figsize=(10, 5))

# Multi bars
# plt.bar(x + (0- (n-1)/2)*width, movie_a, width, label='Movie A')
# plt.bar(x + (1- (n-1)/2)*width, movie_b, width, label='Movie B')
# plt.bar(x + (2- (n-1)/2)*width, movie_c, width, label='Movie C')
# plt.bar(x + (3- (n-1)/2)*width, movie_d, width, label='Movie D')

for i, (movie, revenue) in enumerate(movies.items()):
    offset = (i - (n-1)/2) * width
    plt.bar(x + offset, revenue, width, label = movie)

# Labels
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.title("Movie Revenue Comparison")

# X ticks
plt.xticks(x, years)

# Legend
plt.legend()

# Grid
plt.grid(axis='y')

plt.show()
```

# Stack Bar chart

```python
years = [2020, 2021, 2022]

movie_a = [500, 700, 900]
movie_b = [400, 650, 850]

plt.bar(years, movie_a)
plt.bar(years, movie_b, bottom=movie_a)
```
