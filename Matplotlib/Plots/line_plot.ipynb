```python
import numpy as np
import matplotlib.pyplot as plt

import matplotlib.ticker as ticker


fig, axs = plt.subplots(2, 1, figsize=(5.4, 5.4), layout='constrained')
print(fig, axs)
x = np.arange(100)
for nn, ax in enumerate(axs):
    print(nn, ax)
    # ax.plot(x, x)
    if nn == 1:
        ax.set_title('Manual ticks')
        ax.set_yticks(np.arange(0, 100.1, 100/3))
        xticks = np.arange(0.50, 101, 20)
        xlabels = [f'\\${x:1.2f}' for x in xticks]
        ax.set_xticks(xticks, labels=xlabels)
    # else:
    #     ax.set_title('Automatic ticks')
```

```python
movies = ["Avatar", "Titanic", "Avengers: Endgame", "Frozen"]
years = [2009, 1997, 2019, 2013]
revenues = [2923, 2264, 2799, 1280]  # in millions USD
labels = [f"{movie}\n({year})" for movie, year in zip(movies, years)]

#style
print(plt.style.available)
plt.style.use("seaborn-v0_8-ticks")

# Plot line chart
plt.figure(figsize=(10, 5))
plt.plot(labels, revenues, marker='o')
plt.title("Movie Revenue Comparison")
plt.xlabel("Movies")
plt.ylabel("Revenue (Million USD)")


# Rotate labels if needed
plt.xticks(rotation=15)
# Grid
plt.grid(True)

# Show chart
plt.tight_layout()
plt.show()
```

```python
movies = ["Avatar", "Titanic", "Avengers: Endgame", "Frozen"]
years = [2009, 1997, 2019, 2013]
revenues = [2923, 2264, 2799, 1280]  # million USD

plt.style.use("default")
plt.figure(figsize=(10, 5))

plt.plot(years, revenues, marker='o')

# Add movie names near points
for x, y, movie in zip(years, revenues, movies):
    plt.text(x, y, movie, fontsize=9, ha='right')

# Labels and title
plt.xlabel("Year")
plt.ylabel("Revenue (Million USD)")
plt.title("Movie Revenue Over Years")

# Grid
plt.grid(True)

plt.show()
```
