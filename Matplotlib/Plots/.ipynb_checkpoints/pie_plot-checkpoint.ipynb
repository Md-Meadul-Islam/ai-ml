# Pie Chart
## What percentage does each category contribute to the whole?

Examples:
- market share
- budget allocation
- revenue share

  
| Option | Purpose | Example |
|---|---|---|
| `x` | slice sizes | `[100, 200]` |
| `labels` | slice names | `["Salary", "R&D"]` |
| `autopct` | show percentage | `"%1.1f%%"` |
| `colors` | slice colors | `["red", "green"]` |
| `explode` | pull out slice | `[0, 0.2]` |
| `shadow` | add shadow | `True` |
| `startangle` | rotate pie | `90` |
| `radius` | pie size | `1.2` |
| `counterclock` | direction | `True` |
| `wedgeprops` | slice styling | `{"edgecolor":"black", "linewidth": 2, "linestyle":"--", "fill":False}` `dict(width=0.4)` |
| `textprops` | text styling | `{"fontsize":12, "color": "blue"}` |

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
expences = ["Salaries","R&D","Rent","Marketing","Miscellaneous"]
amounts = [400, 300, 200, 100, 250]

plt.axis('equal') #Without this, the pie may look oval.
plt.pie(amounts, labels=expences, autopct="%1.2f%%", explode = [0,0.1, 0, .2, 0.01], shadow=True)
plt.show()
```

```python
movies = ["Avatar", "Titanic", "Frozen", "Batman"]
revenue = [40, 30, 20, 10]

plt.pie(revenue, labels=movies, wedgeprops=dict(width=0.6))
plt.axis('equal')
plt.show()
```

```python
plt.pie(revenue, labels=movies, wedgeprops={"edgecolor":"black", "linewidth": 1,"linestyle":"--", "animated":True})
plt.axis('equal')
plt.show()
```
