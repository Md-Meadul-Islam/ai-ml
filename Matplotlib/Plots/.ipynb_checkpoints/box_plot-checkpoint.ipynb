# Box Plot

<table style="width: 100%; border-collapse: collapse; border: none;">
<tr>
    
## What is the spread, center, and outliers of the data?

Examples:
- salary analysis
- exam score spread
- performance comparison
</tr>
<tr>

<td style="vertical-align: top; width: 50%; padding-right: 15px; border: none;">

| Option | Purpose | Example |
|---|---|---|
| `x` | numerical data | `[10, 20, 30]` |
| `tick_labels` | category names | `["Class A"]` |
| `vert` | vertical/horizontal | `True` |
| `patch_artist` | fill box color | `True` |
| `showmeans` | show mean value | `True` |
| `showfliers` | show outliers | `True` |
| `widths` | box width | `0.5` |
| `notch` | notch around median | `True` |
| `whis` | whisker length rule | `1.5` |
| `medianprops` | median line styling | `{"color":"red"}` |
| `boxprops` | box styling | `{"linewidth":2}` |
| `whiskerprops` | whisker styling | `{"linestyle":"--"}` |
| `capprops` | cap styling | `{"linewidth":2}` |
| `flierprops` | outlier styling | `{"marker":"o"}` |

</td>

<td style="vertical-align: top; width: 50%; padding-left: 15px; border: none;">

| Part | Meaning | Calculation Formula | Note |
|---|---|---|---|
| Minimum | smallest value | `Q1 − 1.5×IQR` | `IQR = Q3 − Q1` |
| Q1 | 25th percentile | Lower half median | start → middle |
| Median (Q2) | middle value | — | center |
| Q3 | 75th percentile | Upper half median | middle → end |
| Maximum | largest value | `Q3 + 1.5×IQR` | upper bound |

</td>

</tr>
</table>

```python
import matplotlib.pyplot as plt
```

```python
marks = [10, 15, 20, 22, 25, 27, 30,31,32, 100, 110]
plt.boxplot(marks)
plt.grid(True)
```

```python
marks = [10, 15, 20, 22, 25, 27, 30]
plt.boxplot(marks, showmeans=True, showfliers=True)
plt.grid(True)
```

```python
marks = [20, 25, 30, 35, 40, 45, 50, 55, 60, 90]

plt.boxplot(
    marks,
    patch_artist=True,
    showmeans=True,
    notch=True,
    widths=0.5
)

plt.title("Student Marks Distribution")

plt.show()
```

```python
class_a = [20, 30, 40, 50, 60]
class_b = [35, 45, 55, 65, 75]

plt.boxplot([class_a, class_b], tick_labels=["A", "B"], patch_artist=True, showmeans=True)

plt.show()
```

```python
plt.boxplot(
    marks,
    patch_artist=True,
    boxprops=dict(facecolor='skyblue')
)
```

```python
math = [40,50,60,70,80]
science = [45,55,65,75,85]

plt.boxplot(
    [math, science],
    tick_labels=['Math','Science']
)

plt.show()
```
