## How are values distributed across ranges?

Examples:
- age distribution
- salary distribution
- exam marks frequency
  
</td>
<td>

| Option | Purpose | Example |
|---|---|---|
| `x` | numerical data | `[10, 20, 30]` |
| `bins` | number of ranges/groups | `5` |
| `color` | bar color | `"blue"` |
| `edgecolor` | border color | `"black"` |
| `linewidth` | border thickness | `2` |
| `alpha` | transparency | `0.7` |
| `label` | legend name | `"Marks"` |
| `histtype` | histogram style | `"bar"` |
| `orientation` | vertical/horizontal | `"horizontal"` |
| `rwidth` | relative bar width | `0.8` |
| `density` | normalize frequency | `True` |
| `stacked` | stacked histograms | `True` |

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(3284973014)
scores = np.random.normal(70, 10, 1000)
```

```python
plt.hist(scores, bins = 50, edgecolor="blue", density=True)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")

plt.show()
```

```python
scores2 = np.random.normal(10, 40, 1000)
plt.hist(scores, bins = 50, label="Score 1", edgecolor="blue", density=True)
plt.hist(scores2, bins = 50, label="Score 2", edgecolor="red", density=True, alpha=0.5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)
plt.title("Marks Distribution")

plt.show()
```
