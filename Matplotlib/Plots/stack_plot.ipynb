```python
import matplotlib.pyplot as plt
```

```python
years = [2020, 2021, 2022, 2023]

movie_a = [10, 20, 10, 40]
movie_b = [15, 25, 30, 30]
movie_c = [5, 10, 15, 20]

plt.stackplot(
    years,
    movie_a,
    movie_b,
    movie_c
)
plt.grid(True)
plt.show()
```

```python
plt.stackplot(
    years,
    movie_a,
    movie_b,
    movie_c,
    labels=['Movie A', 'Movie B', 'Movie C'],
    colors=['skyblue', 'orange', 'green']
)

plt.legend()
plt.show()
```

```python
years = [2020, 2021, 2022, 2023]

movie_a = [10, 20, 30, 40]
movie_b = [15, 25, 20, 30]
movie_c = [5, 10, 15, 20]

plt.figure(figsize=(8,5))

plt.stackplot(
    years,
    movie_a,
    movie_b,
    movie_c,
    labels=['Movie A', 'Movie B', 'Movie C'],
    colors=['skyblue', 'orange', 'green'],
)

plt.xlabel("Year")
plt.ylabel("Revenue")
plt.title("Movie Revenue Stack Plot")

plt.legend(loc='upper left')

plt.show()
```
