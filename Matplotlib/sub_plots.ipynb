```python
import matplotlib.pyplot as plt
import numpy as np
```

# SubPlot

```python
x = [i for i in range(1, 11)]

y1 = [np.sqrt(i) for i in x] #square root
y2 = [i*2 for i in x] #double
y3 = [i**2 for i in x] #square
y4 = [i**3 for i in x] #cube

plt.subplot(2, 2, 1) #rows, cols, index
plt.plot(x, y1)
plt.title("Plot_1 - Square root")

plt.subplot(2, 2, 2) #rows, cols, index
plt.plot(x, y2)
plt.title("Plot_2 - Double")

plt.subplot(2, 2, 3) #rows, cols, index
plt.plot(x, y3)
plt.title("Plot_3 - Square")

plt.subplot(2, 2, 4) #rows, cols, index
plt.plot(x, y4)
plt.title("Plot_4 - Cube")

plt.tight_layout()
```

# SubPlots (Modern Matplotlib)

```python
fig, axis = plt.subplots(2, 2, figsize = (5,5)) #rows, cols, canvas size(inch)
```

```python
x = [i for i in range(1, 11)]

y1 = [np.sqrt(i) for i in x] #square root
y2 = [i*2 for i in x] #double
y3 = [i**2 for i in x] #square
y4 = [i**3 for i in x] #cube

fig, axis = plt.subplots(2, 2)
axis[0][0].plot(x, y1)
axis[0][0].set_title("Square Root")

axis[0][1].plot(x, y2)
axis[0][1].set_title("Double")

axis[1][0].plot(x, y3)
axis[1][0].set_title("Square")

axis[1][1].plot(x, y4)
axis[1][1].set_title("Cube")

fig.suptitle("Comparing Numbers")
fig.tight_layout()
```

```python
x = [i for i in range(1, 11)]

y1 = [np.sqrt(i) for i in x]
y2 = [i*2 for i in x]
y3 = [i**2 for i in x]
y4 = [i**3 for i in x]

ys = [y1, y2, y3, y4]
titles = ["Square Root", "Double", "Square", "Cube"]

rows = 2
cols = 2

fig, axis = plt.subplots(rows, cols, figsize=(6,6))

k = 0

for i in range(rows):
    for j in range(cols):
        axis[i][j].plot(x, ys[k])
        axis[i][j].set_title(titles[k])
        k += 1

plt.tight_layout()
plt.show()
```

```python
data = {
    "Square Root": y1,
    "Double": y2,
    "Square": y3,
    "Cube": y4
}

rows = 2
cols = 2

fig, axis = plt.subplots(rows, cols, figsize=(6, 6))

axis = axis.flatten()  # convert 2D array → 1D for easy looping

for ax, (title, y) in zip(axis, data.items()):
    print(ax)
    ax.plot(x, y)
    ax.set_title(title)

plt.tight_layout()
plt.show()
```
