## LOF - Local Outlier Factor


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("thyroid_dataset.csv")
```

```python
X = df.drop("Outlier_label", axis=1)
y = df["Outlier_label"]
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.neighbors import LocalOutlierFactor

neighbours = LocalOutlierFactor(contamination=0.0361)

labels = neighbours.fit_predict(X_scaled)
```

```python
from sklearn.decomposition import PCA

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(10,8))
plt.scatter(x=X_pca[:,0], y=X_pca[:, 1], c= labels)
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.title("Anomalies in Thyroid datasets")
```

```python
import numpy as np

outliers = np.sum(labels==-1)
normal = np.sum(labels==1)
print(outliers, normal)
```
